# Load the Clean **Data**

```python
import pandas as pd

# Load the clean dataset generated from File 1
df = pd.read_csv('clean_dataset.csv')
print("Dataset Shape:", df.shape)
df.head()
```

# Train the Winning Random Forest **Model**

```python
import joblib
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler

# 1. Drop any rows containing missing values
df_clean = df.dropna().copy()

# 2. Separate Features (X) and Target (y)
X = df_clean.drop(columns=['isFraud'])
y = df_clean['isFraud'].astype(int)

# 3. Fit scaler on full clean feature set
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# 4. Train final Random Forest model on full dataset
final_model = RandomForestClassifier(n_estimators=50, random_state=42, n_jobs=-1)
final_model.fit(X_scaled, y)

print("Final Random Forest model trained successfully!")
```

# **Export model.pkl and scaler.pkl**

```python
from google.colab import files

# Save serialized binary files
joblib.dump(final_model, 'model.pkl')
joblib.dump(scaler, 'scaler.pkl')

print("Saving complete! Triggering file downloads...")

# Download model and scaler to your computer
files.download('model.pkl')
files.download('scaler.pkl')
```
